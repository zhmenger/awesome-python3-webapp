# 计算机基础知识与编程概念

## 一、计算机常识

### 计算机硬件系统五大组成部分

计算机硬件系统由以下五个核心部分组成：

1. **运算器**
2. **控制器**
3. **存储器**
4. **输入设备**
5. **输出设备**

### 核心组件详解

#### 中央处理器（CPU）
- **组成**：计算器 + 控制器
- **功能**：执行各种运算和控制指令

#### 存储器
- **分类**：
  - 内部存储器（内存）
  - 外部存储器
- **特点**：中央处理器可以直接寻址的存储空间

#### I/O设备
- **输入设备**：键盘、鼠标、麦克风、摄像头
- **输出设备**：显示器、打印机、扬声器

### 冯·诺依曼体系关键点

1. 将存储器和中央处理器分开
2. 将数据以二进制方式编码

## 二、变量和类型

### 变量定义

- **概念**：用来保存数据的内存空间
- **特性**：
  - 变量的值可以被读取和修改
  - 是所有运算和控制的基础
  - 在编程语言中是数据的载体

### 变量命名规则

1. 变量由字母、数字和下划线构成，数字不能开头
2. Python是大小写敏感的编程语言（A 和 a 是两个不同的变量）
3. 变量名不能跟Python的关键字重名，尽量避开Python的保留字
   - **关键字示例**：`is`、`if`、`else`、`for`、`while`、`True`、`False`等
   - **保留字示例**：`int`、`print`、`input`、`str`、`math`、`os`等

### 命名惯例

1. 变量名通常用小写英文字母，多个单词用下划线进行连接
2. 受保护的变量用单个下划线开头
3. 私有的变量用两个下划线开头

```python
# 数据类型
# 1. 整型（int）
print(0b100)  # 二进制整数
print(0o100)  # 八进制整数
print(100)    # 十进制整数
print(0x100)  # 十六进制整数

# 2.浮点型（float）
print(123.456)    # 数学写法
print(1.23456e2)  # 科学计数法

# 3.字符串型（str）
print('Hello')  # 单引号包裹
print("Hello")  # 双引号包裹

# 4.布尔型（bool），布尔型只有True、False两种值, 布尔值可以用 and, or, not运输

# 5.空值，基本上每种编程语言都有自己的特殊值--空值，在Python中，用None表示
```

## 二、变量的使用

```python
# 变量使用的案例，使用变量保存数据并进行加减乘除运算

a = 45
b = 12
print(a, b)    #  45 12
print(a + b)   #  57
print(a - b)   #  33
print(a * b)   #  540
print(a / b)   #  3.75
```

在Python中可以用`type`函数对变量进行检查，程序中的函数概念与数据上的函数概念非常类似，包含函数名、自变量和因变量

```python
# 使用type函数检查变量的类型

a = 100
b = 123.45
c = 'Hello, world'
d = True

print(type(a))   # <class 'int'>
print(type(b))   # <class 'float'>
print(type(c))   # <class 'str'>
print(type(d))   # <class 'bool'>
```

可以通过Python内置函数来改变变量类型，常用变量相关函数
- `int()`: 将一个数值或者字符串转换成整数，可以指定进制。
- `float()`: 将一个字符串（在可能的情况下）转换成浮点数。
- `str()`: 将指定对象转换成字符串形式，可以指定编码方式
- `chr()`: 将整数（字符编码）转换成对应的（一个字符）字符串
- `ord()`: 将（一个字符的）字符串转换成对应的整数（字符编码）
- `bool()`: 将指定对象转换成布尔类型，输出True或者False

```python
# 变量的类型转换操作

a = 100
b = 123.45
c = '123'
d = '100'
e = '123.45'
f = 'Hello, world'
g = True

print(float(a))         # int类型的100转成float，100.0
print(int(b))           # float类型的123.45转成int，输出123
print(int(c))           # str类型的123.45转成int，输出123
print(int(c, base=16))  # 类型的'123'按十六进制转成int，输出291
print(int(d, base=2))   # str类型的'100'按二进制转成int，输出4
print(float(e))         # str类型的'123.45'转成float，输出123.45
print(bool(e))          # str类型的'hello, world'转成bool，输出True
print(int(g))           # bool类型的True转成int，输出1
print(chr(a))           # int类型的100转成str，输出'd'
print(ord('d'))         # str类型的'd'转成int，输出100
```

## 基本数据类型转换
Python中基本数据类型转换的方法有几个。

| **方法** | **说明** |
|:-------|:--------|
| int(x [,base ]) | 将x转换为一个整数 | 
| float(x ) | 将x转换到一个浮点数 | 
| complex(real [,imag ]) |创建一个复数 | 
| str(x ) | 将对象 x 转换为字符串 | 
| repr(x ) | 将对象 x 转换为表达式字符串 |
| eval(str ) | 用来计算在字符串中的有效Python表达式,并返回一个对象 |
| tuple(s )	| 将序列 s 转换为一个元组 |
| list(s ) | 将序列 s 转换为一个列表 |
| chr(x ) | 将一个整数转换为一个字符 |
| unichr(x ) | 将一个整数转换为 Unicode 字符 |
| ord(x ) | 将一个字符转换为它的整数值 |
| hex(x ) | 将一个整数转换为一个十六进制字符串 |
| oct(x ) | 将一个整数转换为一个八进制字符串 |
